## 🕳️ Bloque 3 — Valores nulos

Los valores nulos representan información faltante. Pueden ocurrir porque el dato no fue recolectado, no aplica, o hubo un error.

Hay dos estrategias principales:
- **Eliminar** las filas o columnas con nulos
- **Imputar** (completar) los nulos con un valor representativo


### Ejercicio 3.1 — Cuantificar nulos por columna

```python
# Nulos por columna, ordenados de mayor a menor
df.isna().sum()
```
