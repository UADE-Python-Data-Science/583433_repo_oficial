### Ejercicio 2.2 — Convertir `price` a numérico

La columna `price` viene como texto con el símbolo `$` y comas.
Hay que limpiarla antes de convertirla a número.

**Pasos:**
1. Remover el `$` con `.str.replace()`
2. Remover las `,` con `.str.replace()`
3. Convertir a float con `pd.to_numeric()`


```python
# Cast con astype
df['price'].astype("int", errors="ignore") # si no puede convertir -> no cambia el valor

# Cast con to_numeric
pd.to_numeric(df['price'], errors='coerce') # si no puede convertir -> convierte a NaN
```

```python
# Paso 1: remover el símbolo $
df['price'] = df['price'].str.replace("$", "", regex=False)

# Paso 2: remover las comas
df['price'] = df['price'].str.replace(",", "", regex=False)

# Paso 3: convertir a numérico
df['price'] = pd.to_numeric(df['price'], errors='coerce')

print(f"dtype de price: {df['price'].dtype}")
df['price'].head()
```

```python
df["price"].isnull().sum()
```

```python
df['last_review'].isnull().sum()
```

```python
# Convertí last_review a datetime
df['last_review'] = pd.to_datetime(df["last_review"], errors="coerce")

print(f"dtype de last_review: {df['last_review'].dtype}")
df['last_review'].dt.year.head()
```
