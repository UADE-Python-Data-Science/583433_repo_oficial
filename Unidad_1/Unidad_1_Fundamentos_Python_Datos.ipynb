## 1. Variables y tipos de datos primitivos

Una **variable** es un nombre que usamos para referenciar un valor guardado en memoria. En Python no hace falta declarar el tipo de dato: el intérprete lo infiere automáticamente a partir del valor asignado (esto se llama *tipado dinámico*).

Los **tipos de datos primitivos** más usados son:

| Tipo | Nombre en Python | Ejemplo |
|---|---|---|
| Texto | `str` | `"Tacos El Gordo"` |
| Entero | `int` | `4` |
| Decimal | `float` | `4.6` |
| Booleano | `bool` | `True` / `False` |
| Vacío / nulo | `NoneType` | `None` |

Vamos a declarar algunas variables usando la primera fila de nuestro dataset de tacos.

````python
# Datos de la primera taquería del dataset
nombre = "Tacos El Gordo"
score = 4.6
latitud = 32.6305
longitud = -117.0601
tiene_reviews = True
sucursal_cerrada = None

print(nombre, score, latitud, longitud, tiene_reviews, sucursal_cerrada)
````

````python
# La función type() nos dice el tipo de dato de una variable
print(type(nombre))
print(type(score))
print(type(tiene_reviews))
print(type(sucursal_cerrada))
````

### Conversión de tipos (*casting*)

Muchas veces los datos llegan en un formato que no es el que necesitamos (por ejemplo, un número guardado como texto). Python nos permite convertir entre tipos con `int()`, `float()`, `str()` y `bool()`.

````python
# El score llega como texto (por ejemplo, desde un archivo o un formulario)
score_texto = "4.6"
print(type(score_texto))

# Lo convertimos a float para poder operar matemáticamente con él
score_numero = float(score_texto)
print(type(score_numero))
print(score_numero + 0.1)  # ahora sí podemos sumarle un número
````

````python
# ¡Ojo! No todas las conversiones son válidas
nombre = "Tacos El Gordo"
# float(nombre)  # Esto rompe: ValueError, porque "Tacos El Gordo" no es un número

# bool() interpreta 0, "", None, [] y {} como False. Cualquier otro valor no vacío es True
print(bool(0))
print(bool(4.6))
print(bool(""))
print(bool("Tacos El Gordo"))
````

### f-strings

Para armar mensajes que combinan texto con variables, usamos **f-strings**: se escriben con el prefijo `f` antes de las comillas, y las variables van entre llaves `{}`.

````python
nombre = "Tacos El Gordo"
score = 4.6

print(f"{nombre} tiene un puntaje de {score} sobre 5")
print(f"{nombre} tiene un puntaje de {score:.0f} sobre 5")  # formateamos sin decimales
````

## 2. Listas

Una **lista** es una colección **ordenada** y **mutable** de elementos. Se define con corchetes `[]`, y sus elementos pueden ser de cualquier tipo (incluso mezclados). Al estar ordenados, accedemos a cada elemento a través de un **índice**, que arranca en `0`.

````python
# Lista con los nombres de las primeras 5 taquerías del dataset
taquerias = ["Tacos El Gordo", "The Taco Stand", "Las Cuatro Milpas", "Lucha Libre Taco Shop", "Oscars Mexican Seafood"]
print(taquerias)
print(len(taquerias))  # cantidad de elementos
````

````python
# Accedemos por índice
print(taquerias[0])   # primer elemento
print(taquerias[-1])  # último elemento

# Slicing: sublistas
print(taquerias[0:2])  # los dos primeros
print(taquerias[2:])   # desde el tercero hasta el final
````

### Métodos más usados

| Método | Qué hace |
|---|---|
| `.append(x)` | agrega `x` al final |
| `.insert(i, x)` | inserta `x` en la posición `i` |
| `.remove(x)` | elimina la primera aparición de `x` |
| `.pop(i)` | elimina y devuelve el elemento en la posición `i` (por defecto, el último) |
| `.sort()` | ordena la lista (in place) |
| `.index(x)` | devuelve la posición de `x` |
| `.count(x)` | cuenta cuántas veces aparece `x` |

````python
scores = [4.6, 4.7, 4.8, 4.3, 4.7]

scores.append(4.6)          # agregamos el score de "Lola 55"
print("Después de append:", scores)

scores.sort(reverse=True)   # ordenamos de mayor a menor
print("Ordenado:", scores)

print("Score más alto:", scores[0])
print("Promedio:", sum(scores) / len(scores))
````

## 3. Diccionarios

Un **diccionario** almacena información en pares `clave: valor`. Se define con llaves `{}`. A diferencia de las listas, no accedemos por índice sino por **clave**, y son ideales para representar una **entidad** con varios atributos (por ejemplo, una taquería con su nombre, score y ubicación).

````python
# Representamos una única taquería como diccionario
taqueria = {
    "id": "SDT-001",
    "nombre": "Tacos El Gordo",
    "score": 4.6,
    "latitud": 32.6305,
    "longitud": -117.0601,
    "review": "The adobada is non-negotiable. Authenticity at its peak."
}

print(taqueria["nombre"])
print(taqueria["score"])
````

### Métodos más usados

| Método | Qué hace |
|---|---|
| `.get(clave)` | obtiene el valor (devuelve `None` en vez de error si no existe) |
| `.keys()` | devuelve todas las claves |
| `.values()` | devuelve todos los valores |
| `.items()` | devuelve pares `(clave, valor)` |
| `.update({...})` | agrega o modifica pares |
| `.pop(clave)` | elimina una clave y devuelve su valor |

````python
# .get() es más seguro que [] cuando no estamos seguros de que la clave exista
print(taqueria.get("nombre"))
print(taqueria.get("telefono"))          # no existe -> devuelve None
print(taqueria.get("telefono", "N/A"))   # podemos definir un valor por defecto
````

````python
# Agregamos y modificamos información
taqueria["review_count"] = 1204          # agregamos un nuevo par clave-valor
taqueria["score"] = 4.7                  # actualizamos el score
print(taqueria)

for clave, valor in taqueria.items():
    print(f"{clave}: {valor}")
````

## 4. Estructuras de datos complejas

Hasta ahora vimos listas y diccionarios "simples". Pero en la práctica, los datos casi siempre se combinan: **listas de listas**, **listas de diccionarios** y **diccionarios de listas**. Entender bien estas tres combinaciones es clave, porque son la base conceptual de lo que más adelante va a ser un `DataFrame` de `pandas`.

### 4.1 Listas de listas (matrices)

Una lista puede contener otras listas como elementos. Esto nos permite representar **datos tabulares**: cada sublista es una "fila".

````python
# Cada fila: [nombre, score, latitud, longitud]
matriz_taquerias = [
    ["Tacos El Gordo", 4.6, 32.6305, -117.0601],
    ["The Taco Stand", 4.7, 32.8427, -117.2731],
    ["Las Cuatro Milpas", 4.8, 32.7003, -117.1431],
]

print(matriz_taquerias[0])       # primera fila completa
print(matriz_taquerias[0][0])    # nombre de la primera fila
print(matriz_taquerias[2][1])    # score de la tercera fila (Las Cuatro Milpas)
````

````python
# Recorrer una matriz: fila por fila, y dentro de cada fila, columna por columna
for fila in matriz_taquerias:
    nombre, score, lat, lon = fila   # unpacking: desempaquetamos la fila en 4 variables
    print(f"{nombre}: score {score}, ubicado en ({lat}, {lon})")
````

**Limitación:** en una lista de listas, tenemos que recordar qué representa cada posición (`fila[0]` es el nombre, `fila[1]` es el score...). Esto es frágil y poco legible. Por eso, cuando cada fila representa una entidad con atributos con nombre, conviene usar una **lista de diccionarios**.

### 4.2 Listas de diccionarios

## 5. Condicionales

Los condicionales permiten ejecutar distintos bloques de código según se cumpla o no una condición. Se construyen con `if`, `elif` (opcional, puede haber varios) y `else` (opcional).

Los **operadores de comparación** (`==`, `!=`, `>`, `<`, `>=`, `<=`) devuelven un `bool`, y se pueden combinar con `and`, `or` y `not`.

````python
score = 4.6

if score >= 4.7:
    categoria = "Excelente"
elif score >= 4.5:
    categoria = "Muy bueno"
elif score >= 4.0:
    categoria = "Bueno"
else:
    categoria = "Regular"

print(f"Score {score} -> {categoria}")
````

````python
# Combinando condiciones con and / or
score = 4.4
lat = 32.8475

# ¿Es un taco de "excelente" calidad Y está al norte de San Diego (lat > 32.8)?
if score >= 4.5 and lat > 32.8:
    print("Excelente y al norte de la ciudad")
elif score >= 4.5 or lat > 32.8:
    print("Cumple al menos una condición")
else:
    print("No cumple ninguna condición")
````

## 6. Estructuras de control: `while` y `for`

### `for`

El bucle `for` recorre los elementos de un iterable (una lista, un diccionario, un string, un rango de números...) uno por uno. Es la estructura que más vamos a usar para recorrer datos.

````python
# Recorremos la lista de diccionarios y clasificamos cada taquería
for taco in tacos:
    if taco["score"] >= 4.7:
        clasificacion = "Excelente"
    elif taco["score"] >= 4.5:
        clasificacion = "Muy bueno"
    else:
        clasificacion = "Bueno"
    print(f"{taco['nombre']}: {clasificacion}")
````

````python
# enumerate() nos da el índice y el valor a la vez
for i, taco in enumerate(tacos):
    print(f"{i}: {taco['nombre']}")
````

````python
# range() genera una secuencia de números, muy útil para iterar un número fijo de veces
for i in range(5):
    print(f"Iteración número {i}")
````

### `while`

El bucle `while` repite un bloque de código **mientras** se cumpla una condición. A diferencia del `for`, no sabemos de antemano cuántas veces se va a ejecutar, así que hay que tener cuidado de que la condición eventualmente se vuelva falsa (si no, entramos en un bucle infinito).

````python
# Vamos sumando scores hasta superar 20 puntos acumulados, ¿cuántas taquerías necesitamos?
i = 0
acumulado = 0

while acumulado < 20:
    acumulado += tacos[i]["score"]
    i += 1

print(f"Con las primeras {i} taquerías superamos 20 puntos acumulados (total: {acumulado:.1f})")
````

### `break` y `continue`

- `break` corta el bucle inmediatamente.
- `continue` salta a la siguiente iteración, sin ejecutar el resto del bloque para el elemento actual.

````python
# break: nos quedamos con el primer taco con score >= 4.8
for taco in tacos:
    if taco["score"] >= 4.8:
        print("Encontrado:", taco["nombre"])
        break

# continue: saltamos los tacos con score menor a 4.6
for taco in tacos:
    if taco["score"] < 4.6:
        continue
    print(taco["nombre"], taco["score"])
````

## 7. Funciones

Una **función** encapsula un bloque de código reutilizable. Se define con `def`, puede recibir parámetros (con valores por defecto opcionales) y puede retornar un valor con `return`. Encapsular lógica en funciones evita repetir código y hace que los programas sean más fáciles de leer, testear y mantener.

````python
def clasificar_score(score):
    """Clasifica el score de una taquería en una categoría textual."""
    if score >= 4.7:
        return "Excelente"
    elif score >= 4.5:
        return "Muy bueno"
    elif score >= 4.0:
        return "Bueno"
    else:
        return "Regular"

print(clasificar_score(4.6))
print(clasificar_score(4.8))
````

````python
# Parámetros con valores por defecto
def resumen_taqueria(taco, incluir_ubicacion=False):
    """Arma un string resumen de una taquería (diccionario) del dataset."""
    texto = f"{taco['nombre']} - {clasificar_score(taco['score'])} ({taco['score']})"
    if incluir_ubicacion:
        texto += f" | ({taco['lat']}, {taco['lon']})"
    return texto

print(resumen_taqueria(tacos[0]))
print(resumen_taqueria(tacos[0], incluir_ubicacion=True))
````

````python
# Usamos la función dentro de un for, para procesar todo el dataset
for taco in tacos:
    print(resumen_taqueria(taco))
````

## 8. Listas por comprensión y funciones `lambda`

### Listas por comprensión

Es una forma compacta de crear una lista nueva aplicando una operación (y, opcionalmente, un filtro) a cada elemento de un iterable, en una sola línea.

```python
[expresion for elemento in iterable if condicion]
```

````python
# Forma tradicional con for
nombres = []
for taco in tacos:
    nombres.append(taco["nombre"])
print(nombres)

# La misma lógica, como lista por comprensión
nombres = [taco["nombre"] for taco in tacos]
print(nombres)
````

````python
# Lista por comprensión CON filtro (if)
excelentes = [taco["nombre"] for taco in tacos if taco["score"] >= 4.7]
print(excelentes)

# También podemos armar diccionarios por comprensión
scores_por_nombre = {taco["nombre"]: taco["score"] for taco in tacos}
print(scores_por_nombre)
````

### Funciones `lambda`

Son funciones **anónimas** de una sola línea, muy útiles cuando necesitamos una función simple como argumento de otra función (por ejemplo, el parámetro `key` de `sorted()`).

```python
lambda argumentos: expresion
```

````python
# Función lambda simple
duplicar = lambda x: x * 2
print(duplicar(4.6))

# El uso más típico: como criterio de orden en sorted()
tacos_ordenados = sorted(tacos, key=lambda taco: taco["score"], reverse=True)
for taco in tacos_ordenados:
    print(taco["nombre"], taco["score"])
````

````python
# map() y filter() combinados con lambda
scores = list(map(lambda taco: taco["score"], tacos))
print("Todos los scores:", scores)

top_tacos = list(filter(lambda taco: taco["score"] >= 4.6, tacos))
print("Cantidad de tacos top:", len(top_tacos))
````

## 9. Expresiones regulares

Las **expresiones regulares** (*regex*) son patrones que describen texto, y sirven para buscar, extraer, validar o reemplazar información dentro de un string. En Python se usan a través del módulo `re`.

Funciones principales:

| Función | Qué hace |
|---|---|
| `re.search(patron, texto)` | busca la **primera** coincidencia en cualquier parte del texto |
| `re.findall(patron, texto)` | devuelve **todas** las coincidencias como lista |
| `re.sub(patron, reemplazo, texto)` | reemplaza las coincidencias por otro texto |
| `re.match(patron, texto)` | busca la coincidencia solo al **inicio** del texto |

Algunos elementos básicos de patrones: `\d` (dígito), `\w` (letra/número/_), `\s` (espacio), `+` (uno o más), `*` (cero o más), `[A-Z]` (rango de caracteres), `.` (cualquier caracter).

````python
import re

review = '"The adobada is non-negotiable. Authenticity at its peak."'

# Las reviews del dataset vienen con comillas dobles al inicio y al final.
# Las eliminamos con re.sub, reemplazando comillas por vacío
review_limpia = re.sub(r'"', '', review)
print(review_limpia)
````

````python
# Extraemos todas las palabras que empiezan con mayúscula (posibles nombres propios o inicios de oración)
texto = "The adobada is non-negotiable. Authenticity at its peak."
palabras_mayuscula = re.findall(r'\b[A-Z][a-zA-Z]*\b', texto)
print(palabras_mayuscula)
````

````python
# Buscamos si una review menciona la palabra "fish" (sin importar mayúsc/minúsc)
reviews = [
    "Incredible smoked fish and spicy shrimp tacos.",
    "The adobada is non-negotiable. Authenticity at its peak.",
    "Unbeatable $1.50 fish tacos and fresh ceviche.",
]

for r in reviews:
    if re.search(r'fish', r, re.IGNORECASE):
        print("Menciona pescado ->", r)
````

````python
# Extraemos montos en dólares de una review, ej: "$1.50"
review = "Unbeatable $1.50 fish tacos and fresh ceviche."
montos = re.findall(r'\$\d+\.\d+', review)
print(montos)
````

## 10. Manejo de excepciones

Cuando algo sale mal durante la ejecución de un programa, Python lanza una **excepción** (un error). Si no la manejamos, el programa se corta. El bloque `try / except` nos permite capturar esos errores y decidir qué hacer, en lugar de que el programa se rompa.

```python
try:
    # código que puede fallar
except TipoDeError:
    # qué hacer si falla
else:
    # se ejecuta si NO hubo error
finally:
    # se ejecuta siempre, haya habido error o no
```

Algunas excepciones comunes: `KeyError` (clave inexistente en un dict), `IndexError` (índice fuera de rango en una lista), `ValueError` (conversión de tipo inválida), `ZeroDivisionError` (división por cero), `TypeError` (operación entre tipos incompatibles).

````python
# Sin manejo de excepciones, esto rompe el programa
taco = {"nombre": "Tacos El Gordo", "score": 4.6}

try:
    print(taco["telefono"])
except KeyError:
    print("La clave 'telefono' no existe en este registro")
````

````python
def convertir_score(valor):
    """Intenta convertir un valor a float. Devuelve None si no es posible."""
    try:
        return float(valor)
    except ValueError:
        print(f"No se pudo convertir '{valor}' a número")
        return None

print(convertir_score("4.6"))
print(convertir_score("cuatro coma seis"))
````

````python
# Podemos capturar varios tipos de error, y usar else / finally
datos = ["4.6", "4.7", "no disponible", "4.8"]

scores_validos = []
for dato in datos:
    try:
        score = float(dato)
    except ValueError:
        print(f"Dato inválido, se descarta: '{dato}'")
    else:
        scores_validos.append(score)
    finally:
        print(f"Se procesó el dato: '{dato}'")

print("Scores válidos:", scores_validos)
````

## 11. Buenas prácticas de escritura de código

Escribir código que funciona es el primer paso; escribir código **legible y mantenible** es lo que distingue a un buen desarrollador. Vamos a ver tres pilares:

### 11.1 PEP8

[PEP8](https://peps.python.org/pep-0008/) es la guía de estilo oficial de Python. Algunas reglas clave:

- Nombres de variables y funciones en `snake_case` (minúsculas, separadas por guión bajo): `score_promedio`, no `scorePromedio` ni `ScorePromedio`.
- Nombres de clases en `PascalCase`: `TaqueriaDataset`.
- Constantes en `MAYUSCULAS_CON_GUIONES`: `SCORE_MINIMO = 4.0`.
- Máximo 79-99 caracteres por línea (según la configuración del equipo).
- Un espacio alrededor de operadores: `a = b + c`, no `a=b+c`.
- Dos líneas en blanco antes de una definición de función a nivel de módulo.
- Nombres descriptivos: `taquerias_excelentes` en vez de `lst` o `x`.

### 11.2 Docstrings

Un **docstring** es un string de documentación que va como primera línea del cuerpo de una función, clase o módulo, entre comillas triples. A diferencia de un comentario, se puede consultar en tiempo de ejecución con `help()` y es leído por herramientas de documentación automática.

### 11.3 Comentarios efectivos

Los comentarios (`#`) deben explicar el **por qué**, no el **qué** (el código ya dice qué hace). Evitá comentarios redundantes u obsoletos: un comentario que no coincide con el código es peor que no tener comentario.

````python
# ---- MAL ----
def f(l):
  r=[]
  for x in l:
    if x['score']>=4.7: # si el score es mayor o igual a 4.7 (comentario redundante)
      r.append(x['nombre'])
  return r

# ---- BIEN ----
SCORE_EXCELENTE = 4.7  # constante en mayúsculas, fácil de ajustar sin buscar "números mágicos" en el código


def obtener_taquerias_excelentes(lista_taquerias):
    """Filtra las taquerías con score excelente.

    Args:
        lista_taquerias (list[dict]): lista de diccionarios, cada uno con al menos la clave "score".

    Returns:
        list[str]: nombres de las taquerías con score >= SCORE_EXCELENTE.
    """
    nombres_excelentes = []
    for taqueria in lista_taquerias:
        if taqueria["score"] >= SCORE_EXCELENTE:
            nombres_excelentes.append(taqueria["nombre"])
    return nombres_excelentes


print(obtener_taquerias_excelentes(tacos))
help(obtener_taquerias_excelentes)  # el docstring queda disponible en tiempo de ejecución
````
