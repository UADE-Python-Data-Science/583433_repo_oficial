# Unidad 1: Fundamentos de Py

```python
import numpy as np
```

```python
# CRear una lista de diccionarios con los siguientes datos
unis = [
    {"nombre": "Stanford", "ranking": 4, "emp_ratio": 97.8},
    {"nombre": "MIT", "ranking": 1, "emp_ratio": 96.2},
    {"nombre": "Yale", "ranking": 9, "emp_ratio": 93.3},
]
```

```python
def mostrar_unis(unis, n):
    n_unis = unis[:n]
    for u in n_unis:
        print(u)

mostrar_unis(unis, 5)
```

```python
def mostrar_unis(unis, cant_reg=None):
    if cant_reg and cant_reg <= len(unis):
        selected_unis = unis[:cant_reg]
    else:
        selected_unis = unis
    for u in selected_unis:
        print(u)

mostrar_unis(unis, 6)
```

```python
# Calificar universidades
def calificar_unis(unis):
    for u in unis:
        if 1 <= u["ranking"] <=5:
            u["nivel"] = "Top5"
        elif 5 < u["ranking"] <=10:
            u["nivel"] = "Top10"
        else:
            u["nivel"] = "Top50"

calificar_unis(unis)
mostrar_unis(unis, 6)
         
```

```python
# Generar una funcion que retorne el diccionario de la universidad con mayor ranking
def buscar_mejor_emp_ratio(unis):
    mayor_emp_rate = unis[0]
    for u in unis:
        if u["emp_ratio"] > mayor_emp_rate["emp_ratio"]:
            mayor_emp_rate = u
    return mayor_emp_rate

print(buscar_mejor_emp_ratio(unis))
```

```python
def calcular_emp_ratio_avg(unis):
    emp_list = [u["emp_ratio"] for u in unis]
    emp_ratio_avg = sum(emp_list) / len(emp_list)
    return emp_ratio_avg

print(calcular_emp_ratio_avg(unis))
```

```python
def uni_mejor_emp_ratio(unis):
    try:
        calificar_unis(unis)
        uni_mr = buscar_mejor_emp_ratio(unis)
        er_avg = calcular_emp_ratio_avg(unis)
        print(
            f"La Universidad con mayor ratio de empleabilidad es {uni_mr["nombre"]} que se encuentra en el {uni_mr["nivel"]} con un índice de empleabilidad de {uni_mr["emp_ratio"]} % sobre un promedio de {er_avg}"
        )
    except Exception as e:
        print(f"ERROR: {e}")
```

```python
uni_mejor_emp_ratio(unis)
```

# Unidad 2: NumPy, Pandas, IDE

```python
def cal_est_emp_unis(unis):
    emp_list = [u["emp_ratio"] for u in unis]
    emp_array = np.array(emp_list)
    count = len(emp_array)
    max = np.max(emp_array)
    min = np.min(emp_array)
    mean = np.mean(emp_array)
    median = np.median(emp_array)
    std = np.std(emp_array)

    print(f"Cantidad: {count}\nMáximo: {max}\nMínimo: {min}\nMedia: {mean:.2f}\nMediana: {median:.2f}\nDesvío: {std:.2f}")

cal_est_emp_unis(unis)
```

# 10. Metrica recomendación

Tomamos 
* `national_rank` : The 2026 Forbes ranking position (1-50) 
[ 1 Mayor Importancia - 0 Menor Importancia]
* `employment_rate`: % of graduates employed or in grad school within 1 year.

```python
df["ranking"] = 51 - df["national_rank"]

df["ranking_norm"] = (
    df["ranking"].max() - df["ranking"]
) / (
    df["ranking"].max() - df["ranking"].min()
)

df["employment_norm"] = (
    df["employment_rate"] - df["employment_rate"].min()
) / (
    df["employment_rate"].max() - df["employment_rate"].min()
)
df["recommendation_index"] = (
    0.6 * df["ranking_norm"] +
    0.4 * df["employment_norm"]
)
recommendation = df.sort_values(
    by="recommendation_index",
    ascending=False
).head(5)

recommendation[["university_name", "national_rank", "employment_rate", "recommendation_index"]]
```
