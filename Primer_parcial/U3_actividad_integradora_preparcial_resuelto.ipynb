### Ejercicio 1.1 — Listas y funciones

Completá la función que, dada una lista de precios, retorna el precio mínimo, máximo y promedio como una tupla.

````python
def analizar_precios(lista_precios):
    """Retorna (minimo, maximo, promedio) de una lista de precios."""
    minimo   = max(lista_precios)
    maximo   = min(lista_precios)
    promedio = round(sum(lista_precios)/len(lista_precios), 2)
    return minimo, maximo, promedio

# Probá con los primeros 10 precios del dataset
precios_muestra = df["price"].head(10).tolist()
print("Precios:", precios_muestra)

minimo, maximo, promedio = analizar_precios(precios_muestra)
print(f"Mínimo: {minimo} | Máximo: {maximo} | Promedio: {promedio}")
````

### Ejercicio 1.2 — Diccionarios

Completá el código para construir un diccionario que cuente cuántos diamantes hay de cada tipo de corte (`cut`), sin usar `value_counts()`.

### Ejercicio 1.3 — Función con condicional

Escribí una función `clasificar_diamante` que reciba el precio de un diamante y retorne su categoría:
- `"Económico"` si price < 1.000
- `"Medio"` si price está entre 1.000 y 5.000
- `"Premium"` si price > 5.000

````python
def clasificar_diamante(precio):
    if precio < 1000:
        return "Económico"
    elif precio < 5000:
        return "Medio"
    else:
        return "Premiun"

# Probá la función con tres precios distintos
print(clasificar_diamante(500))
print(clasificar_diamante(3000))
print(clasificar_diamante(12000))
````

### Ejercicio 1.4 — Manejo de excepciones

Completá la función para que maneje el caso en que el valor recibido no sea numérico.

````python
def precio_seguro(valor):
    """Convierte un valor a float. Si falla, retorna None."""
    try:
        return float(valor)
    except Exception as e:
        return None

print(precio_seguro("1234.5"))    # → 1234.5
print(precio_seguro("abc"))       # → None
print(precio_seguro(None))        # → None
````

### Ejercicio 6.1 — Resumen por corte

Completá la función `resumen_por_corte` que recibe el DataFrame y retorna un diccionario con el siguiente formato para cada tipo de corte:

```python
{
    "Ideal": {
        "cantidad":    21.551,
        "precio_medio": 3.457.54,
        "carat_medio":  0.70
    },
    ...
}
```


````python
def resumen_por_corte(df):
    """
    Retorna un diccionario con estadísticas por tipo de corte.
    Claves: nombre del corte
    Valores: dict con cantidad, precio_medio y carat_medio
    """
    resultado = {}
    cortes_unicos = df["cut"].unique()

    for corte in cortes_unicos:
        # Filtrá el dataframe para ese corte
        subset = df[df["cut"] == corte]

        resultado[corte] = {
            "cantidad":     len(subset),
            "precio_medio": round(subset["price"].mean(), 2),
            "carat_medio":  round(subset["carat"].mean(), 2)
        }

    return resultado

# Ejecutá la función y mostrá el resultado
resumen = resumen_por_corte(df)
for corte, stats in resumen.items():
    print(f"{corte:12} → cantidad: {stats['cantidad']:>6,} | precio medio: ${stats['precio_medio']:>9,.2f} | carat medio: {stats['carat_medio']:.2f}")
````
