### 🧪 Cómo funcionan los tests

En los Bloques 1 y 2 **cada ejercicio es una función**. Debajo de cada función hay una celda de test (**no la modifiques**) que la prueba con varios casos, incluyendo datos distintos de `paises`.

1. Completá los `___` de la función y ejecutá la celda.
2. Ejecutá la celda de test que está debajo.

| Resultado | Significado |
|---|---|
| ✅ `OK` | La función pasó todos los casos |
| ❌ `FALLÓ` | La función corre pero devuelve un resultado incorrecto (el mensaje dice qué caso falló) |
| ⚠️ `ERROR` | La función rompe al ejecutarse. `NameError: name '___'` significa que quedan blancos sin completar |

Ejecutá primero esta celda, que define la herramienta de testing:

```python
import copy

# Hace que un ___ sin completar dé un error claro (en Jupyter, ___ ya existe como variable)
class _Blanco:
    def _falta(self, *args, **kwargs):
        raise NotImplementedError("quedan blancos ___ sin completar")
    for _m in ["__add__", "__radd__", "__sub__", "__rsub__", "__mul__", "__truediv__", "__gt__", "__ge__",
               "__lt__", "__le__", "__eq__", "__ne__", "__bool__", "__call__", "__getitem__", "__len__",
               "__iter__", "__index__", "__hash__", "__float__", "__int__", "__format__", "__getattr__"]:
        locals()[_m] = _falta
    def __repr__(self):
        return "___"

___ = _Blanco()

resultados_tests = {}

def correr_test(test):
    """Ejecuta una función de test y muestra el resultado."""
    nombre = test.__name__
    try:
        test()
        resultados_tests[nombre] = True
        print(f"✅ {nombre}: OK")
    except AssertionError as e:
        resultados_tests[nombre] = False
        print(f"❌ {nombre}: FALLÓ → {e}")
    except NotImplementedError:
        resultados_tests[nombre] = False
        print(f"⚠️ {nombre}: la función todavía tiene blancos ___ sin completar")
    except NameError as e:
        resultados_tests[nombre] = False
        print(f"⚠️ {nombre}: {e} → ejecutá la celda de la función y revisá que no tenga errores de sintaxis")
    except Exception as e:
        resultados_tests[nombre] = False
        print(f"⚠️ {nombre}: ERROR al ejecutar → {type(e).__name__}: {e}")
```
