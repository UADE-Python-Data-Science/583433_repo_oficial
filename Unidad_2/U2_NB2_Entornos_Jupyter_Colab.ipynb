### ¿Qué es una celda?

Un notebook se compone de **celdas**. Cada celda puede ser:

| Tipo | Qué hace | Cómo ejecutar |
|---|---|---|
| **Code** | Ejecuta código Python | `Shift + Enter` |
| **Markdown** | Muestra texto con formato | `Shift + Enter` |
| **Raw** | Texto sin procesar (poco usado) | — |

### Atajos de teclado más importantes

| Atajo | Acción |
|---|---|
| `Shift + Enter` | Ejecutar celda y avanzar |
| `Ctrl + Enter` | Ejecutar celda sin avanzar |
| `A` (modo comando) | Insertar celda arriba |
| `B` (modo comando) | Insertar celda abajo |
| `M` (modo comando) | Convertir a Markdown |
| `Y` (modo comando) | Convertir a Code |
| `D D` (modo comando) | Eliminar celda |
| `Esc` | Entrar al modo comando |
| `Enter` | Entrar al modo edición |

> 💡 **Modo comando vs. modo edición:** presionás `Esc` para salir de la celda (modo comando, borde azul) y `Enter` para editarla (modo edición, borde verde).

## 3. Markdown — escribir con formato en los notebooks

### 3.1 Sintaxis básica de Markdown

A continuación, cada bloque explica la sintaxis y muestra el resultado al renderizarse:

---

**Títulos** (con `#`):
```

## Título nivel 2

```

**Énfasis:**
```

**Listas:**
```
- Ítem 1
- Ítem 2
  - Sub-ítem

1. Primero
2. Segundo
3. Tercero
```

**Links e imágenes:**
```
[texto del link](https://url.com)

```

**Bloques de código:**
```
```python
import pandas as pd
df = pd.DataFrame()
```
```

**Tablas:**
```
| Columna A | Columna B |
|-----------|-----------|
| Valor 1   | Valor 2   |
```

**Línea horizontal:**
```
---
```

**Citas:**
```
> Esto es una cita o nota destacada.
```

### 🚀 Actividad 4 — Creá tu análisis desde cero

Seguí los pasos en orden. Cada paso corresponde a un bloque de código que vas a escribir en una celda nueva.


### Paso 1 — Importar librerías y crear el DataFrame

Copiá y ejecutá en tu notebook:


````python
# Paso 1 — Librerías y datos
import pandas as pd

# Lista de diccionarios → misma estructura del NB3
tacos = [
    {"ID": "SDT-001", "Restaurant": "Tacos El Gordo",        "Score": 4.6, "Latitude": 32.6305, "Longitude": -117.0601},
    {"ID": "SDT-002", "Restaurant": "The Taco Stand",         "Score": 4.7, "Latitude": 32.8427, "Longitude": -117.2731},
    {"ID": "SDT-003", "Restaurant": "Las Cuatro Milpas",      "Score": 4.8, "Latitude": 32.7003, "Longitude": -117.1431},
    {"ID": "SDT-004", "Restaurant": "Lucha Libre Taco Shop",  "Score": 4.3, "Latitude": 32.7441, "Longitude": -117.1724},
    {"ID": "SDT-005", "Restaurant": "Oscars Mexican Seafood", "Score": 4.7, "Latitude": 32.8028, "Longitude": -117.2536}
]

df = pd.DataFrame(tacos)
print(f"DataFrame creado: {df.shape[0]} filas × {df.shape[1]} columnas")
df
````

### Paso 2 — Exploración inicial

Probá cada uno de estos comandos **en celdas separadas**:

````python
# Información general del DataFrame
df.info()
````

````python
# Estadísticas descriptivas
df.describe()
````

````python
# Primeras 5 filas
df.head()
````

````python
# Muestra aleatoria de 4 filas
df.sample(4, random_state=42)
````

### Paso 3 — Selección de datos

````python
# Seleccionar una columna → Series
print(df["Restaurant"])
````

````python
# display(df)
# print(df)
df
````

````python
# Seleccionar múltiples columnas → DataFrame
df[["Restaurant", "Score"]]
````

````python
# .iloc — selección por posición (fila 0 a 2, columna 0 a 2)
df.iloc[0:3, 3:]
````

### Paso 4 — Filtrado

````python
# Restaurantes con score >= 4.7
top = df[df["Score"] >= 4.7]
print(f"Restaurantes top (score >= 4.7): {len(top)}")
top[["Restaurant", "Score"]]
````

````python
# Score alto Y al norte (Latitude > 32.75)
norte_top = df[(df["Score"] >= 4.6) & (df["Latitude"] > 32.75)]
norte_top[["Restaurant", "Score", "Latitude"]]
````

### Paso 5 — Crear columna y ordenar

````python
# Nueva columna: categoría basada en score
df["Categoria"] = df["Score"].apply(
    lambda x: "⭐ Excelente" if x >= 4.7 else "👍 Muy bueno"
)

# Ordenar de mayor a menor score
df.sort_values("Score", ascending=False)[["Restaurant", "Score", "Categoria"]]
````

### Paso 6 — Groupby y estadísticas por grupo

````python
# ¿Cuántos restaurantes hay en cada categoría y cuál es el score promedio?
resumen = df.groupby("Categoria")["Score"].agg(
    cantidad="count",
    promedio="mean",
    maximo="max"
)
print(resumen)
````

### Paso 7 — Visualización rápida

````python
import matplotlib.pyplot as plt

# Distribución de scores
df["Score"].value_counts().sort_index().plot(
    kind="bar",
    color="coral",
    edgecolor="black",
    figsize=(8, 4)
)
plt.title("Distribución de scores — Top Tacos San Diego")
plt.xlabel("Score")
plt.ylabel("Cantidad de restaurantes")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

## 5. Jupyter vs. Google Colab — ¿cuándo usar cada uno?

| Característica | Jupyter (local) | Google Colab |
|---|---|---|
| **Instalación** | Requiere instalar Python + Jupyter | Solo necesitás un navegador |
| **Acceso** | Tu máquina | Desde cualquier dispositivo |
| **Librerías** | Las instalás vos con `pip` | pandas, numpy, sklearn ya incluidos |
| **GPU/TPU** | Depende de tu hardware | Gratis (con límites de tiempo) |
| **Persistencia de datos** | En tu disco local | En Google Drive (si guardás) |
| **Privacidad** | Todo local, sin conexión | Los datos se suben a servidores de Google |
| **Colaboración** | Compartir el archivo `.ipynb` | Link para editar en tiempo real |
| **Archivos externos** | Acceso directo al sistema de archivos | Hay que subir los archivos o montar Drive |
| **Velocidad** | Depende de tu CPU/RAM | Generalmente buena para notebooks medianos |
| **Precio** | Gratuito (open source) | Gratuito (con limitaciones) / Colab Pro pago |

### ¿Cuándo usar Jupyter local?

- Cuando trabajás con **datos sensibles** que no podés subir a la nube
- Cuando necesitás **acceso al sistema de archivos** (bases de datos locales, múltiples archivos)
- Cuando querés trabajar **sin conexión a internet**
- En proyectos profesionales o de investigación con datasets grandes

### ¿Cuándo usar Google Colab?

- Para **aprender y explorar** rápidamente sin configurar nada
- Cuando necesitás **GPU** para modelos de deep learning
- Para **compartir notebooks** con compañeros o docentes fácilmente
- Para trabajar desde una computadora que no es la tuya
