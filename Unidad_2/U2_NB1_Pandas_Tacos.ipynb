En este notebook vamos a explorar sus estructuras fundamentales usando un dataset real: los mejores taquerías de San Diego, California. 🌮


```python
import pandas as pd
import numpy as np
print(f'pandas version: {pd.__version__}')
```

---
## 1. Series — la estructura unidimensional

Una **Series** es un array unidimensional con un índice etiquetado:

```python
lista = [4.6, 4.7, 4.8, 4.3, 4.7]
lista[2]
```

```python
scores = pd.Series([4.6, 4.7, 4.8, 4.3, 4.7], index=["s1", "s2", "s3","s4","s5"])
scores["s3"]
```

```python
# Series desde una lista
scores = pd.Series([4.6, 4.7, 4.8, 4.3, 4.7],
                   index=["El Gordo", "The Taco Stand", "Las Cuatro Milpas",
                          "Lucha Libre", "Oscars Seafood"],
                   name="score")
print(scores)
print("\ndtype:", scores.dtype)
print("índice:", scores.index.tolist())
```

```python
# Acceso por etiqueta y por posición
print("El Gordo:", scores["El Gordo"])
print("Top 3:\n", scores[:3])

# Operaciones vectorizadas (igual que NumPy)
print("\nScores + 0.1:\n", scores + 0.1)
print("\nRestaurantes con score >= 4.7:\n", scores[scores >= 4.7])
```

---
## 2. DataFrame — la estructura tabular

Un **DataFrame** es una tabla de datos con filas y columnas etiquetadas.  
Vamos a crearlo desde la lista de diccionarios del dataset de tacos:

```python
# El dataset completo viene de una lista de diccionarios
# (generada a partir del Excel top_tacos_in_san_diego.xlsx)
tacos_list = [
  {
    "ID": "SDT-001",
    "Restaurant Name": "Tacos El Gordo",
    "Score": 4.6,
    "Latitude": 32.6305,
    "Longitude": -117.0601,
    "Review Snippet": "The adobada is non-negotiable. Authenticity at its peak."
  },
  {
    "ID": "SDT-002",
    "Restaurant Name": "The Taco Stand",
    "Score": 4.7,
    "Latitude": 32.8427,
    "Longitude": -117.2731,
    "Review Snippet": "Handmade tortillas make the Al Pastor truly stand out."
  },
  {
    "ID": "SDT-003",
    "Restaurant Name": "Las Cuatro Milpas",
    "Score": 4.8,
    "Latitude": 32.7003,
    "Longitude": -117.1431,
    "Review Snippet": "Old school charm with the best chorizo in Barrio Logan."
  },
  {
    "ID": "SDT-004",
    "Restaurant Name": "Lucha Libre Taco Shop",
    "Score": 4.3,
    "Latitude": 32.7441,
    "Longitude": -117.1724,
    "Review Snippet": "Get the Undefeated Taco; the salsa bar is top-tier."
  }
]

# Convertir lista de diccionarios a DataFrame
df = pd.DataFrame(tacos_list)
print("Tipo:", type(df))
print("Shape:", df.shape)   # (filas, columnas)
df
```

---
## 3. Exploración inicial del DataFrame

Los primeros comandos que ejecutamos al recibir un nuevo dataset:

```python
# .info() — tipos de datos, valores nulos, memoria
df.info()
```

```python
# .head() y .tail() — primeras y últimas filas
print('--- head(3) ---')
print(df.head())
print('\n--- tail(3) ---')
print(df.tail(3))
```

```python
# .sample() — muestra aleatoria (útil para exploración rápida)
df.sample(2, random_state=42)
```

```python
# .describe() — estadísticas descriptivas de columnas numéricas
df.describe(include="object")
```

```python
# .dtypes — tipo de dato de cada columna
df.dtypes
```

---
## 4. Selección de datos

```python
# Seleccionar una columna → Series
print(df["Restaurant Name"].head())
print("\nTipo:", type(df["Restaurant Name"]))
```

```python
# Seleccionar múltiples columnas → DataFrame
df[["Restaurant Name", "Score"]].head()
```

```python
# .loc[] — selección por etiqueta de índice
print(df.loc[0:2, ["Restaurant Name", "Score"]])
```

```python
# .iloc[] — selección por posición numérica
print(df.iloc[0:3, 0:3])
```

---
## 5. Filtrado y consultas

```python
# Filtrar restaurantes con score mayor o igual a 4.7
top_rated = df[df["Score"] >= 4.7]
print(f"Restaurantes con score >= 4.7: {len(top_rated)}")
top_rated[["Restaurant Name", "Score"]].sort_values("Score", ascending=False)
```

```python
# Múltiples condiciones: score alto Y latitud al norte de San Diego centro
norte_y_top = df[(df["Score"] >= 4.6) & (df["Latitude"] > 32.78)]
norte_y_top[["Restaurant Name", "Score", "Latitude"]]
```

---
## 6. Operaciones sobre columnas

```python
# Crear una nueva columna: Score redondeado a entero (categoría)
df["Score_Category"] = df["Score"].apply(lambda x: "🌟 Excelente" if x >= 4.7 else "👍 Muy bueno")
df[["Restaurant Name", "Score", "Score_Category"]].head(8)
```

```python
# Estadísticas por grupo
print("Estadísticas por categoría:")
print(df.groupby("Score_Category")["Score"].agg(["count", "mean", "min", "max"]))
```

---
## 7. Ordenamiento

```python
# Ordenar por score descendente
df.sort_values("Score", ascending=False)[["Restaurant Name", "Score"]].head(10)
```

---
## 8. Exportación en múltiples formatos

```python
import os
os.makedirs("/mnt/user-data/outputs", exist_ok=True)

# CSV
df.to_csv("/mnt/user-data/outputs/tacos_sd.csv", index=False)

# JSON
df.to_json("/mnt/user-data/outputs/tacos_sd.json", orient="records", indent=2, force_ascii=False)

# Excel
df.to_excel("/mnt/user-data/outputs/tacos_sd_export.xlsx", index=False)

print("Archivos exportados:")
print("  ✅ tacos_sd.csv")
print("  ✅ tacos_sd.json")
print("  ✅ tacos_sd_export.xlsx")
```

---
## 9. Visualización rápida con pandas

Pandas integra matplotlib para visualizaciones directas:

```python
import matplotlib.pyplot as plt

# Distribución de scores
df["Score"].value_counts().sort_index().plot(kind="bar", figsize=(8, 4), color="coral", edgecolor="black")
plt.title("Distribución de scores — Top Tacos San Diego")
plt.xlabel("Score")
plt.ylabel("Cantidad de restaurantes")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 🔑 Resumen de comandos explorados

| Comando | Descripción |
|---|---|
| `pd.DataFrame(lista_de_dicts)` | Crear DataFrame desde lista de diccionarios |
| `df.info()` | Tipos, nulos y memoria |
| `df.head(n)` | Primeras n filas |

| `df.describe()` | Estadísticas descriptivas |
| `df["columna"]` | Seleccionar una columna (Series) |
| `df[["col1", "col2"]]` | Seleccionar varias columnas |
| `df.loc[filas, cols]` | Selección por etiqueta |
| `df.iloc[filas, cols]` | Selección por posición |
| `df[condicion]` | Filtrado booleano |
| `df.sort_values("col")` | Ordenar por columna |
