## ¿Por qué usar entornos virtuales?

Cuando trabajamos en varios proyectos de Python, cada uno puede necesitar versiones distintas de las mismas librerías. Por ejemplo:
- El Proyecto A necesita `pandas==1.5.0`
- El Proyecto B necesita `pandas==2.1.0`

Si instalamos todo globalmente, generamos conflictos. Un **entorno virtual** es una instalación aislada de Python y sus librerías, exclusiva para un proyecto.

> 💡 **Regla de oro:** Un proyecto = un entorno virtual.


---
## 1. Crear un entorno virtual

### 🪟 Windows (cmd o PowerShell)

```bash
# Crear el entorno (se crea una carpeta llamada 'venv')
python -m venv venv

# Activar el entorno
venv\Scripts\activate

# El prompt cambia a:
# (venv) C:\tu\proyecto>
```

### 🍎 macOS / Linux (terminal)

```bash
# Crear el entorno
python3 -m venv venv

# Activar el entorno
source venv/bin/activate

# El prompt cambia a:
# (venv) tu@maquina:~/proyecto$
```

> ⚠️ El nombre `venv` es convencional, pero podés usar cualquier nombre.


---
## 2. Instalar librerías dentro del entorno

Una vez activado el entorno, todo lo que instalés queda aislado en él:

```bash
pip install numpy pandas matplotlib jupyter
```

Para ver qué está instalado:

```bash
pip list
```


---
## 3. Guardar las dependencias: `requirements.txt`

Para que otros (o vos en otra máquina) puedan replicar el entorno exacto:

```bash
# Exportar las dependencias con sus versiones
pip freeze > requirements.txt

# Instalar desde el archivo
pip install -r requirements.txt
```

Ejemplo de `requirements.txt`:
```
numpy==1.26.4
pandas==2.2.1
matplotlib==3.8.3
jupyter==1.0.0
```


---
## 4. Desactivar y eliminar el entorno

```bash
# Desactivar (vuelve al Python global)
deactivate

# Eliminar el entorno (simplemente borrar la carpeta)
# Windows:
rmdir /s /q venv

# macOS / Linux:
rm -rf venv
```


---
## 5. Entornos virtuales en VS Code

VS Code detecta automáticamente los entornos virtuales del proyecto. Para seleccionar uno:

1. Abrí la **paleta de comandos**: `Ctrl+Shift+P` (Win) / `Cmd+Shift+P` (Mac)
2. Escribí: **Python: Select Interpreter**
3. Elegí el entorno `./venv/bin/python` (o `./venv/Scripts/python.exe` en Windows)

> 💡 Una vez seleccionado, VS Code lo usa automáticamente para ejecutar celdas en Jupyter y scripts de Python.

## 6. Diferencias entre Google Colab y entorno local

| | Google Colab | Entorno local (VS Code) |
|---|---|---|
| Librerías | Ya vienen instaladas | Hay que instalarlas con `pip` |
| Entorno virtual | No aplica (entorno efímero) | Recomendado para aislar dependencias |
| Persistencia | Los datos se pierden al cerrar | Todo persiste en disco |
| GPU | Disponible (gratis con límites) | Depende del hardware local |
