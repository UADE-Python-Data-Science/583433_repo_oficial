# Introducción a Merge con Pandas

En este notebook vamos a aprender cómo combinar datasets usando distintos tipos de joins.

<img src="https://www.red-gate.com/simple-talk/wp-content/uploads/2025/09/Full-Outer-Join-featured-image-scaled.jpg" width="600">

```python
import pandas as pd
```

## Creamos los datasets

```python
fact_chart = pd.DataFrame({
    "song_id": [1, 2, 3, 4, 5, 6, 7, 8, 9],
    "chart_id": [101, 101, 102, 102, 103, 103, 104, 104, 105],
    "highest_charting_position": [1, 5, 10, 3, 8, 2, 7, 4, 6],
    "streams": [500000, 300000, 150000, 400000, 250000, 450000, 200000, 350000, 100000],
    "artist_followers": [1000000, 800000, 500000, 1200000, 600000, 1100000, 700000, 900000, 400000]
})

fact_chart.shape
```

```python
dim_songs = pd.DataFrame({
    "song_id": [1, 2, 3, 4, 5, 10, 11, 12],
    "song_name": ["Song A", "Song B", "Song C", "Song D", "Song E", "Song X", "Song Y", "Song Z"],
    "artist": ["Artist 1", "Artist 2", "Artist 3", "Artist 4", "Artist 5", "Artist X", "Artist Y", "Artist Z"],
    "genre": ["Pop", "Rock", "Hip-Hop", "Pop", "Jazz", "Electronic", "Pop", "Rock"],
    "release_date": pd.to_datetime([
        "2020-01-01", "2020-02-01", "2020-03-01", "2020-04-01",
        "2020-05-01", "2021-01-01", "2021-02-01", "2021-03-01"
    ])
})

dim_songs.shape
```

## INNER JOIN

```python
df_inner_join = pd.merge(fact_chart, dim_songs, on="song_id", how="inner")
df_inner_join.shape
```

```python
df_inner_join.sort_values(by="song_id")
```

## LEFT JOIN

```python
df_left_join = pd.merge(fact_chart, dim_songs, on="song_id", how="left")
df_left_join.shape
```

```python
df_left_join.sort_values(by="song_id")
```

## RIGHT JOIN

```python
df_right_join = pd.merge(fact_chart, dim_songs, on="song_id", how="right")
df_right_join.shape
```

```python
df_right_join
```

## FULL OUTER JOIN

* Vemos los datasets combinados
* NaN donde no hay match
* Pero no sabemos fácilmente de dónde vino cada fila

```python
df_outer_join = pd.merge(fact_chart, dim_songs, on="song_id", how="outer")
df_outer_join.shape
```

```python
df_outer_join
```

## Indicador de origen

* Vemos los datasets combinados
* NaN donde no hay match
* Se agrega una columna adicional, "_merge" que indica como se generó la fila

```python
pd.merge(fact_chart, dim_songs, on="song_id", how="outer", indicator=True)
```
